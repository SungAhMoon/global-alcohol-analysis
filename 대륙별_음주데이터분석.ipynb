#### 발표 참고용 (상관관계 설명, 산점도 그래프에 포함될 내용임)

- 상관관계를 분석하여, 각 주류 소비량이 총 알코올 섭취량에 어떠한 영향을 주는지 파악!
- 분석 대상은 국가별 음주 데이터이며, 상관계수 계산을 통해 변수 간 연관성 정도를 정량적으로 확인하였다


1) beer_servings ↔ total_litres_of_pure_alcohol (0.8358)
맥주 제공량과 "국가별 1인당 연간 순수 알코올 소비량" 간 상관계수는 약 0.836(2개의 상관성이 가장 높음), 네 변수 중 가장 높은 수치를 보임. 즉, 맥주 소비가 많은 국가일수록 전체 알코올 섭취량도 높다는 강한 양의 상관관계를 나타냄.

2) wine_servings ↔ total_litres_of_pure_alcohol (0.6676)
와인 제공량과 순수 알코올 소비량 간 상관계수는 약 0.668, 중간 이상의 양의 상관관계를 보인다. 와인 소비가 늘어날수록 순수 알코올 소비량 역시 증가하는 경향 확인 가능.

3) spirit_servings ↔ total_litres_of_pure_alcohol (0.6550)
증류주 제공량과 순수 알코올 소비량 간 상관계수는 약 0.655, 와인과 유사한 수준의 중간 양의 상관관계를 가진다. 증류주 소비량이 많을수록 전체 알코올 섭취량이 높아지는 경향 확인 가능.

4) beer_servings ↔ wine_servings (0.5272)
맥주와 와인 제공량 간 상관계수는 약 0.527로, 보통 수준의 양의 상관관계를 시사한다. 즉, 맥주 소비가 많은 국가일수록 와인 소비도 상대적으로 많은 편.

5) beer_servings ↔ spirit_servings (0.4588)
맥주와 증류주 제공량 간 상관계수는 약 0.459로, 약한 ~ 중간 수준의 양의 상관관계를 보인다. 맥주 소비가 증가하면 증류주 소비도 일부 연동되는 경향이 있으나, 와인과의 관계보다는 약함

6) wine_servings ↔ spirit_servings (0.1948)
와인과 증류주 제공량 간 상관계수는 약 0.195로, 매우 낮은 수준. 이는 와인 소비가 많다고 해서 증류주 소비가 반드시 많지 않음을 의미함. 두 변수 간의 상관관계가 거의 없다는 것을 나타냄.


=> total_litres_of_pure_alcohol 피처가 대체적으로 다른 모든 피처와 강한 상관관계가 있는 것으로 보임
국가별 음주 지표 네 변수 간 피어슨 상관관계를 계산하여, 맥주 제공량이 국가별 1인당 연간 순수 알코올 소비량에 가장 큰 영향을 미치며(0.8358), 그다음으로 와인(0.6676), 증류주(0.6550) 순임을 확인했다.

```python
# 2 - 대륙별 국가 수 비율 - 파이

continents = drinks["continent"].value_counts()
```

```python
continents
```

```python
# 총 서빙 합계 컬럼 생성 - beer_servings + spirit_servings + wine_servings
# total_servings 라는 새로운 열을 형성 한 데이터프레임임

drinks['total_servings'] = ( drinks['beer_servings'] + drinks['spirit_servings'] + drinks['wine_servings'])
```

```python
# 3 - 대륙별 총 서빙량 - 막대 + 평균값
continent_total_servings = (drinks.groupby('continent')['total_servings'].sum().reset_index())
```

```python
continent_total_servings
```

```python
# 총 서빙량의 평균값
# continent_total_servings 데이터 프레임에서 total_servings(3개 알코올 더한)을 평균으로 구한 것
avg_servings = continent_total_servings['total_servings'].mean()
```

```python
avg_servings
```

```python
# 4 - 대륙별 주종별 서빙량 - 막대그래프

# 이건 위에 코드 적어둔거
cor_cols = drinks.columns[1:4]
cor_cols

# 3개의 컬럼을 기준으로 대륙별로 그룹화하여 합계를 구한 것임
continents_sum = drinks[cor_cols].groupby(drinks["continent"]).sum()
```

```python
# 대륙별 주종별 총 서빙량
continents_sum
```

```python
# 대륙별 주종별 총 서빙량의 평균
continents_sum_mean = drinks[cor_cols].groupby(drinks["continent"]).sum().mean()
```

```python
continents_sum_mean
```

### 주종별 가장 높은 대륙

```python
beverage_cols = ['beer_servings', 'spirit_servings', 'wine_servings']
max_continent_by_bev = {}

for col in beverage_cols:
    # 대륙별 총 서빙량 합계(Series)
    continent_sum = drinks.groupby('continent')[col].sum()
    # 합계가 가장 큰 인덱스(=대륙 코드)를 구함
    max_cont = continent_sum.idxmax()
    max_continent_by_bev[col] = max_cont

for col, cont in max_continent_by_bev.items():
     print(f"- {col} 총 서빙량이 가장 많은 대륙: {cont}")
```

```python
alc_mean = drinks["total_litres_of_pure_alcohol"].mean()
```

```python
alc_mean
```

```python
con_mean = drinks.groupby("continent")["total_litres_of_pure_alcohol"].mean()
```

```python
con_mean
```

```python
# 전체 평균 이상인 대륙 (하위 3대륙 - 유럽 > 북미 > 남미)
con_mean[con_mean >= alc_mean]
```

```python
# 전체 평균 이하인 대륙 (하위 3대륙 - 오세아니아 > 아프리카 > 아시아)
con_mean[con_mean <= alc_mean]
```

# 주종별 유럽의 상위 3국가, 하위 3국가

### beer_servings 가 가장 높은 대륙

```python
max_beer = drinks.groupby("continent")["beer_servings"].sum().idxmax() # 최댓값의 맥스를 뽑아줌
```

```python
max_beer
```

```python
# 유럽(EU) 내에서 1인당 맥주 서빙 횟수가 가장 높은 상위 3개 국가
Top3_EU_beer = drinks[drinks["continent"] == max_beer].sort_values(by="beer_servings", ascending=False)[["country", "beer_servings"]].head(3).reset_index(drop=True)
```

```python
Top3_EU_beer
```

```python
Bot3_EU_beer = drinks[drinks["continent"] == max_beer].sort_values(by="beer_servings", ascending=False)[["country", "beer_servings"]].tail(3).reset_index(drop=True)
```
